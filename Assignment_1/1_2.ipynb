```python
import os
from datasets import load_dataset

os.makedirs("raw_data", exist_ok=True)
print("download started")

ds = load_dataset(
    "oscar-corpus/OSCAR-2301", 
    language="hi", 
    split="train", 
    streaming=True
)

count = 0
with open("raw_data/oscar_hi_raw.txt", "w", encoding="utf-8") as f:
    for row in ds:
        text = row["text"].strip()
        if text:
            f.write(text + "\n")
            count += 1
            
            if count >= 520000:
                break

print("Finshed")
```
