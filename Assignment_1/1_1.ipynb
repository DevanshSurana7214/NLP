```python
import os
from datasets import load_dataset

os.makedirs("raw_data", exist_ok=True)
print("download started")

ds = load_dataset(
    "ai4bharat/IndicCorpV2", 
    data_files="data/hi-1.txt", 
    split="train", 
    streaming=True
)

count = 0
with open("raw_data/hin_Deva_raw.txt", "w", encoding="utf-8") as f:
    for row in ds:
        text = row["text"].strip()
        if text:
            f.write(text + "\n")
            count += 1
            
            if count >= 520000:
                break

print("Finshed")
```

```python
import re
import pandas as pd
```

```python
with open("./raw_data/hin_Deva_raw.txt", "r", encoding="utf-8") as f:   
    text=f.read()

print(text[:500])
print(len(text))
```

```python
def sentence_tokenize(text):
    sentences=re.split(r"(?<=[।!?॥])\s+|(?<=[a-zA-Z\u0900-\u097F]\.)\s+", text)
    sentences=[s.strip() for s in sentences if s.strip()]
    return sentences
sentences=sentence_tokenize(text)
```

```python
for i in range(0,15):
    print(sentences[i])
```

```python
token_pattern=re.compile(
    r'https?://\S+|www\.\S+\.\S+' #url
    r'|([\w\.-]+@[\w\.-]+\.\w+)' #mail
    r'|(\d{1,2}[-/]\d{1,2}[-/]\d{2,4})' #dates
    r'|(\d+\.\d+)' #decimal
    r'|(\d+)' #Integers
    r'|([a-zA-Z\u0900-\u0963\u0966-\u097F]+)' #English or Hindi words
    r'|([^\w\s])' #Punctuations

)


def word_tokenize(sentence):
    tokens=[]
    for match in token_pattern.finditer(sentence):
        tokens.append(match.group())

    return tokens

test_text=sentences[0]
print(f"{word_tokenize(test_text)}")
```

```python
df=pd.DataFrame({
    "sentence":sentences
})
df.head()
```

```python
df["tokens"]=df["sentence"].apply(word_tokenize)
df["token_count"]=df["tokens"].apply(len)
df.head()
```

```python
import os
```

```python
os.makedirs("tokenized_output", exist_ok=True)
df.to_parquet(
    "tokenized_output/hin_corpus_tokenized.parquet",
    index=False,
    engine="pyarrow",
    compression="snappy"
)
```

```python
df=pd.read_parquet(
    "tokenized_output/hin_corpus_tokenized.parquet"
)

df.head()
```

```python
all_token_flat=[
    token
    for token_list in df["tokens"]
    for token in token_list
]
```

```python
total_sentences=len(df)
```

```python
total_tokens=df['token_count'].sum()
```

```python
total_characters=sum(len(token) for token in all_token_flat)

```

```python
avg_sentence_len=total_tokens/total_sentences if total_sentences>0 else 0
avg_word_len=total_characters/total_tokens if total_tokens>0 else 0
```

```python
unique_tokens=len(set(all_token_flat))
ttr=unique_tokens/total_tokens if total_tokens>0 else 0
```

```python
print(f"i.   Total number of sentences: {total_sentences:,}")
print(f"ii.  Total number of words: {total_tokens:,}")
print(f"iii. Total number of characters: {total_characters:,}")
print(f"iv.  Average Sentence Length: {avg_sentence_len:.2f} words/sentence")
print(f"v.   Average word length: {avg_word_len:.2f} characters/word")
print(f"vi.  Type/Token Ratio (TTR): {ttr:.6f} ({unique_tokens:,} unique / {total_tokens:,} total)")
```
